```python
#Q1. Let city = ("Mumbai", 19.07, 72.87). Unpack it into three variables name, lat
and lon in a single statement and print them in a sentence. Then swap the values of x = 10 and y = 20 using tuple
assignment (no third variable).


#SOLUTION

# Unpacking the tuple
city = ("Mumbai", 19.07, 72.87)
name, lat, lon = city
print(f"The city of {name} is located at latitude {lat} and longitude {lon}.")

# Swapping x and y using tuple assignment
x = 10
y = 20
print(f"Before swap: x = {x}, y = {y}")

x, y = y, x
print(f"After swap: x = {x}, y = {y}")
```

```python
#Q3. Using a for loop and continue, print the numbers from 1 to 20 skipping all multiples of 4
and also the number 13.


#SOLUTION

# Loop from 1 to 20
for i in range(1, 21):

    # Skip multiples of 4 and the number 13
    if i % 4 == 0 or i == 13:
        continue
    print(i)
```

```python
#Q4 Let x = 10 be a global variable. Write a function that creates a LOCAL x = 20 and prints it, and print
the global x before and after the call. Then write a second function that uses the global keyword to change x to 99.
Comment on both behaviours.


#SOLUTION

# Initialize the global variable
x = 10

def demo_local_scope():
    # Creating a local variable x. This shadows the global x inside this function.
    x = 20
    print(f"Inside demo_local_scope (local): x = {x}")

def demo_global_scope():
    # Using the global keyword allows us to modify the global variable directly
    global x
    x = 99
    print(f"Inside demo_global_scope (modified global): x = {x}")

# 1. Show behavior of local variable shadowing
print(f"Before calling demo_local_scope (global): x = {x}")
demo_local_scope()
print(f"After calling demo_local_scope (global): x = {x}")

print("-" * 40)

# 2. Show behavior of using the global keyword
print(f"Before calling demo_global_scope (global): x = {x}")
demo_global_scope()
print(f"After calling demo_global_scope (global): x = {x}")

# Comment on both behaviours:
# - In demo_local_scope, assigning x = 20 creates a new local variable in the function's scope.
#   The global variable x remains unchanged outside the function.
# - In demo_global_scope, the 'global x' statement tells Python to use the existing global x
#   rather than creating a new local one. Thus, reassigning x changes it globally.
```
